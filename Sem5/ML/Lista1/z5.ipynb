# 1. Histogram 3D
Wygeneruj milion punktów z dwuwymiarowego rozkładu normalnego o średniej (0, 0) i identycznościowej 
macierzy kowariancji używając funkcji numpy.random.randn(...). Narysuj histogram takich danych (wskazówka: histogram będzie wykresem 3D)


```python
import numpy as np
import plotly.graph_objects as go

means = [0.0,0.0]
covariance_matrix = [[1.0,0.0],[0.0,1.0]]
liczby_2d = np.random.multivariate_normal(mean=means,cov=covariance_matrix,size=1000000)
liczby_2d
x= liczby_2d[:,0]
y = liczby_2d[:,1]
z_data, x_edges,y_edges = np.histogram2d(x,y,bins=50,density=True)

fig = go.Figure(data=[go.Surface(
    z=z_data,
    x=x_edges[:-1],
    y=y_edges[:-1],
    opacity=0.5,
    colorscale='Blues'
)])
fig.update_layout(
    title = "3D Histogram of million numbers in a 2d normal distribution",
    scene=dict(
        xaxis_title = "Coordinate X",
        yaxis_title= " Coordinate Y",
        zaxis_title = "Number of points"
    )
)
fig.show(renderer="browser")
```

# 2. Dodanie funkcji gęstości
Na histogramie wygenerowanych danych narysuj funkcję gęstości rozkładu prawdopodobień-
stwa (wskazówka: wykres będzie 3D). Porównaj histogram z funkcją gęstości (wskazówka: wy-
godnie będzie zrobić wykres dynamiczny, umożliwiający obracanie i powiększanie, umożliwia
to na przykład pakiet plotly).


```python
from scipy.stats import multivariate_normal as m_norm

x_f = np.linspace(min(x),max(x),1000)
y_f = np.linspace(min(y),max(y),1000)

Xf,Yf = np.meshgrid(x_f,y_f)

pos = np.dstack((Xf,Yf))

density_f = m_norm(means,covariance_matrix)

Z = density_f.pdf(pos)

fig.add_trace(go.Surface(
    z=Z,
    x=Xf,
    y=Yf,
    colorscale='Reds',
    opacity=0.3,
    name="2D Density function"
))
fig.update_layout(
    title="Comparing Density Function to one million generated numbers in 2D Normal Distribution",
    scene=dict(
        zaxis_title="Density"
    )
)
fig.show(renderer="browser")
```
