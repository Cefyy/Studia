## Inicjalizacja zbioru danych

```python
from ucimlrepo import fetch_ucirepo 
  
# fetch dataset 
iris = fetch_ucirepo(id=53) 
  
# data (as pandas dataframes) 
irysy = iris.data.features 
gatunki = iris.data.targets 
  
# variable information 
print(iris.variables) 
```

## 1. Scatter plot

```python
import plotly.express as px




px.scatter(irysy,x=irysy.index,y="sepal length",labels = {"index": "No. of observation", "sepal length" : "Sepal length in cm"}, title = "Sepal lengths across the observations")
```

#### Czy jest sens łączyć kropki?

##### Nie, ponieważ kolejne obserwacje są niezależne od poprzednich

## 2. Histogram dla iris setosa


```python
import pandas as pd

przedzialy = [0,2.5,3.0,3.5,4.0,4.5,5.0]
nazwy = ['< 2.5',' 2.5 - 3.0',' 3.0 - 3.5',' 3.5 - 4.0',' 4.0 - 4.5',' > 4.5']

print(gatunki['class'].unique()) # jakie są klasy

irysy_setosa = irysy.loc[gatunki['class'] == 'Iris-setosa'].copy() # odfiltrowuje dane tylko dla jednego gatunku
irysy_setosa['sepal widths in setosa'] = pd.cut(irysy_setosa['sepal width'], bins = przedzialy, labels=nazwy, right= False) # dodaje nową kolumnę która przypisuje odpowiedni przedzial danemu sepal width

counts = irysy_setosa['sepal widths in setosa'].value_counts().sort_index().reset_index(name="count")

fig =px.bar(counts,x="sepal widths in setosa",y="count",labels= {"sepal widths in setosa" : "Sepal width [cm]", "count" : "Number of flowers"}, title = "Sepal widths across Setosa species")
fig.show()
```

### Tak, utworzony wykres to histogram, ponieważ wykres przedstawia liczebności

# 3. Scatter plot sepal length i sepal width

```python
irysy_calosc = pd.concat([irysy,gatunki],axis=1)
nowe_nazwy = {
    "Iris-setosa" : "Setosa",
    "Iris-versicolor" : "Versicolor",
    "Iris-virginica" : "Virginica"
}
irysy_calosc["class"] = irysy_calosc["class"].replace(nowe_nazwy)
px.scatter(irysy_calosc,x="sepal length", y="sepal width",color = "class",size="petal length", labels={"sepal length" : "sepal length [cm]", "petal length" : "petal length [cm]", "class" : "Species", "sepal width" : "sepal width [cm]"}, title = "Sepal Width vs. Sepal Length by Iris Species")
```

### Widać, że sepal width gatunku satosa jest widocznie większy, sepal oraz petal length jest mniejszy od pozostałych dwóch gatunków, więc łatwo odróżnić go od pozostałych.
### Pozostałe dwa gatunki mają podobny rozkład szerokości i petal lengths, ale virginica ma zauważalnie dłuższe sepal lengths.

# 4. Pary scatter plotów

```python
fig = px.scatter_matrix(irysy_calosc, dimensions=["sepal length","sepal width","petal length","petal width"],color = "class",title = "Zestawienie wszystkich par atrybutów")

fig.update_traces(diagonal_visible = False)
fig.update_layout(width=1800,height=1800)
fig.show()
```

# 5. Violin plot


```python
fig_petal_l = px.violin(irysy_calosc,x="class",y="petal length",color = "class",points="all")
fig_petal_l.show()

fig_petal_w = px.violin(irysy_calosc,x="class",y="petal width",color = "class",points="all")
fig_petal_w.show()

fig_sepal_l = px.violin(irysy_calosc,x="class",y="sepal length",color = "class",points="all")
fig_sepal_l.show()

fig_sepal_w = px.violin(irysy_calosc,x="class",y="sepal width",color = "class",points="all")
fig_sepal_w.show()
```
