### 1. Histogram rozkładu normalnego

Wygeneruj milion liczb z rozkładu normalnego o średniej 0 i wariancji 1 używając funkcji
numpy.random.randn(...). Narysuj histogram takich danych

```python
import numpy as np
import plotly.express as px
import pandas as pd

liczby = np.random.randn(1000000)

df_liczby = pd.DataFrame(liczby)
fig = px.histogram(df_liczby,opacity=0.6,histnorm='probability density',title="Histogram of million generated values in standard normal distribution")
fig.show()
```

# 2. Funkcja gęstości
Na histogramie wygenerowanych danych narysuj funkcję gęstości rozkładu prawdopodobień-
stwa. Porównaj histogram z funkcją gęstości.

```python
from scipy.stats import norm


x = np.linspace(min(liczby),max(liczby),100)
p = norm.pdf(x,loc=0,scale=1)

fig.add_scatter(
    x=x,
    y=p,
    mode = 'lines',
    line=dict(color='red',width=2),
    name = "Funkcja gęstości"
)
fig.show()
```
