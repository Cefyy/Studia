```python
import numpy as np
from sklearn.cluster import KMeans
from sklearn.mixture import GaussianMixture
import plotly.graph_objects as go
from plotly.subplots import make_subplots




for i in range(1,6):
    X = np.load(f'data{i}.npy')
    
    kmeans = KMeans(n_clusters=3,random_state=42)
    gmm = GaussianMixture(n_components=3,random_state=42)
    
    labels_kmeans = kmeans.fit_predict(X)
    labels_gmm = gmm.fit_predict(X)
    
    
    fig = make_subplots(rows=1,cols=2,subplot_titles=("K-Means","GMM"))
    fig.add_trace(go.Scatter(x=X[:,0],y=X[:,1],mode='markers',marker=dict(color=labels_kmeans, colorscale='Viridis')),
        row=1, col=1)
    fig.add_trace(
        go.Scatter(x=X[:, 0], y=X[:, 1], mode='markers', 
                   marker=dict(color=labels_gmm, colorscale='Viridis')),
        row=1, col=2
    )
    fig.update_layout(title_text=f"Porównanie modeli dla zbioru data_{i}", 
                      showlegend=False, 
                      height=500, width=1000)
    fig.show()
```
