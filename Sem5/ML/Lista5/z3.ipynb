```python
import numpy as np
from sklearn.cluster import KMeans
from scipy.stats import mode
from sklearn.metrics import accuracy_score

X = np.load('image_emb.npy')
labels_data = np.load('image_labels.npy',allow_pickle=True)


known_mask = (labels_data != '')
known_ids = np.where(known_mask)[0]
known_classes = labels_data[known_mask]

kmeans = KMeans(n_clusters=10,random_state=42)

all_clusters = kmeans.fit_predict(X)

cluster_mapping = {}

for cluster_id in range(10):
    mask = (all_clusters[known_ids] == cluster_id)
    classes_in_cluster = known_classes[mask]
    
    
    if len(classes_in_cluster)>0 :
        unique_class,counts = np.unique(classes_in_cluster,return_counts=True)
        most_freq_class = unique_class[np.argmax(counts)]
        cluster_mapping[cluster_id] = most_freq_class
    else:
        cluster_mapping[cluster_id] = -1
        
X_test = np.load('image_emb_test.npy')

y_test_true = np.load('image_labels_test.npy',allow_pickle=True)

test_clusters = kmeans.predict(X_test)

pred_classes = [cluster_mapping.get(cluster,-1) for cluster in test_clusters]

acc = accuracy_score(y_test_true,pred_classes)
print(f"Model accuracy: {acc * 100:.2f}%")  
```
